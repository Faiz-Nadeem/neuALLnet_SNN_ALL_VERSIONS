```python
import os
import pandas as pd

DATA_ROOT = "/kaggle/input/datasets/faiznadeem39/milk10k"

IMAGE_ROOT = os.path.join(DATA_ROOT, "MILK10k_Training_Input")
GT_PATH = os.path.join(DATA_ROOT, "MILK10k_Training_GroundTruth.csv")
META_PATH = os.path.join(DATA_ROOT, "MILK10k_Training_Metadata.csv")

gt = pd.read_csv(GT_PATH)

print("GroundTruth shape:", gt.shape)
print("\nColumns:")
print(gt.columns.tolist())

print("\nFirst 5 rows:")
display(gt.head())

print("\nImage folder examples:")
print(os.listdir(IMAGE_ROOT)[:10])

print("\nMissing values:")
print(gt.isnull().sum())
```

```python
import pandas as pd

metadata = pd.read_csv(META_PATH)

print("Metadata shape:", metadata.shape)

print("\nMetadata columns:")
print(metadata.columns.tolist())

print("\nFirst 5 rows:")
display(metadata.head())

print("\nExample lesion metadata:")
display(
    metadata[
        metadata["lesion_id"] == "IL_0000652"
    ]
)
```

```python
counts = pd.crosstab(metadata["lesion_id"], metadata["image_type"])

valid_lesions = counts[
    (counts["dermoscopic"] > 0) &
    (counts["clinical: close-up"] > 0)
].index

print("Total lesions:", len(counts))
print("Paired lesions:", len(valid_lesions))
print("Missing modalities:", len(counts) - len(valid_lesions))
```

```python
# Create class label from one-hot ground-truth columns
classes = gt.columns.drop("lesion_id")

gt["label"] = gt[classes].idxmax(axis=1)

print("Classes:", classes.tolist())
print("Ground truth shape:", gt.shape)
print(gt["label"].value_counts())
```

```python
# Separate clinical and dermoscopic records
clinical = metadata[
    metadata["image_type"] == "clinical: close-up"
][["lesion_id", "isic_id"]].rename(
    columns={"isic_id": "clinical_id"}
)

dermoscopic = metadata[
    metadata["image_type"] == "dermoscopic"
][["lesion_id", "isic_id"]].rename(
    columns={"isic_id": "dermoscopic_id"}
)

# Pair both modalities using lesion_id
paired = clinical.merge(
    dermoscopic,
    on="lesion_id",
    validate="one_to_one"
)

# Add ground-truth labels
paired = paired.merge(
    gt[["lesion_id", "label"]],
    on="lesion_id",
    validate="one_to_one"
)

# Create image paths
paired["clinical_path"] = paired.apply(
    lambda row: os.path.join(
        IMAGE_ROOT, row["lesion_id"],
        row["clinical_id"] + ".jpg"
    ), axis=1
)

paired["dermoscopic_path"] = paired.apply(
    lambda row: os.path.join(
        IMAGE_ROOT, row["lesion_id"],
        row["dermoscopic_id"] + ".jpg"
    ), axis=1
)

print("Paired dataset shape:", paired.shape)
print("\nClass distribution:")
print(paired["label"].value_counts())

print("\nClinical images exist:",
      paired["clinical_path"].map(os.path.exists).all())

print("Dermoscopic images exist:",
      paired["dermoscopic_path"].map(os.path.exists).all())

display(paired.head())
```

```python
train_parts = []
val_parts = []
test_parts = []

for label in paired["label"].unique():
    class_df = paired[paired["label"] == label].sample(
        frac=1, random_state=42
    )

    n = len(class_df)
    n_test = max(1, round(n * 0.10))
    n_val = max(1, round(n * 0.10))
    n_train = n - n_val - n_test

    train_parts.append(class_df.iloc[:n_train])
    val_parts.append(class_df.iloc[n_train:n_train+n_val])
    test_parts.append(class_df.iloc[n_train+n_val:])

train_df = pd.concat(train_parts).sample(frac=1, random_state=42)
val_df = pd.concat(val_parts).sample(frac=1, random_state=42)
test_df = pd.concat(test_parts).sample(frac=1, random_state=42)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nTotal:", len(train_df) + len(val_df) + len(test_df))
print("\nClass distribution:")
print(pd.DataFrame({
    "Train": train_df["label"].value_counts(),
    "Validation": val_df["label"].value_counts(),
    "Test": test_df["label"].value_counts()
}).fillna(0).astype(int))
```

```python
from torchvision import transforms

IMAGE_SIZE = 224

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomRotation(degrees=20),
    transforms.RandomAffine(
        degrees=0,
        translate=(0.2, 0.2),
        shear=20
    ),
    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.8, 1.0)
    ),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor()
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor()
])

print("Image size:", IMAGE_SIZE)
print("Training augmentation:", train_transform)
print("Evaluation transform:", eval_transform)
```

```python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import torch

class MILK10kDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        clinical = Image.open(
            row["clinical_path"]
        ).convert("RGB")

        dermoscopic = Image.open(
            row["dermoscopic_path"]
        ).convert("RGB")

        if self.transform:
            clinical = self.transform(clinical)
            dermoscopic = self.transform(dermoscopic)

        label = torch.tensor(
            classes.tolist().index(row["label"]),
            dtype=torch.long
        )

        return clinical, dermoscopic, label


print("Dataset class defined successfully.")
```

```python
BATCH_SIZE = 8

train_dataset = MILK10kDataset(train_df, train_transform)
val_dataset = MILK10kDataset(val_df, eval_transform)
test_dataset = MILK10kDataset(test_df, eval_transform)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))
```

```python
import torch
from spikingjelly.activation_based import neuron, surrogate, functional

print("PyTorch:", torch.__version__)
print("SpikingJelly imported successfully")
print("Device:", torch.device("cuda" if torch.cuda.is_available() else "cpu"))
```

```python
import torch
import torch.nn as nn
from spikingjelly.activation_based import neuron, surrogate, functional


# 1. Squeeze-and-Excitation Block
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=4):
        super().__init__()

        mid = max(channels // reduction, 8)

        self.squeeze = nn.AdaptiveAvgPool2d(1)

        self.excitation = nn.Sequential(
            nn.Linear(channels, mid, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(mid, channels, bias=False),
            nn.Sigmoid()
        )

    def forward(self, x):
        b, c, _, _ = x.shape

        weights = self.squeeze(x).view(b, c)
        weights = self.excitation(weights).view(b, c, 1, 1)

        return x * weights


# 2. Depthwise Separable Convolution
class DepthwiseSeparableConv(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()

        self.depthwise = nn.Conv2d(
            in_ch,
            in_ch,
            kernel_size=3,
            padding=1,
            groups=in_ch,
            bias=False
        )

        self.bn_dw = nn.BatchNorm2d(in_ch)

        self.pointwise = nn.Conv2d(
            in_ch,
            out_ch,
            kernel_size=1,
            bias=False
        )

        self.bn_pw = nn.BatchNorm2d(out_ch)

    def forward(self, x):
        x = self.bn_dw(self.depthwise(x))
        x = self.bn_pw(self.pointwise(x))

        return x


# 3. Spiking Block
class SpikeBlock(nn.Module):
    def __init__(self, in_ch, out_ch, tau=2.0):
        super().__init__()

        self.conv = DepthwiseSeparableConv(in_ch, out_ch)

        self.se = SEBlock(out_ch)

        self.lif = neuron.LIFNode(
            tau=tau,
            v_threshold=1.0,
            surrogate_function=surrogate.ATan(alpha=2.0),
            step_mode='s'
        )

        self.pool = nn.MaxPool2d(2, 2)

    def forward(self, x):
        x = self.conv(x)
        x = self.se(x)
        x = self.lif(x)
        x = self.pool(x)

        return x


# 4. NeurALLNet Feature Extractor
class NeurALLFeatureExtractor(nn.Module):
    def __init__(self):
        super().__init__()

        tau = 2.0

        self.stem = nn.Sequential(
            nn.Conv2d(
                3, 48,
                kernel_size=3,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(48),

            neuron.LIFNode(
                tau=tau,
                v_threshold=1.0,
                surrogate_function=surrogate.ATan(alpha=2.0),
                step_mode='s'
            ),

            nn.MaxPool2d(2, 2)
        )

        self.block1 = SpikeBlock(48, 96, tau)
        self.block2 = SpikeBlock(96, 192, tau)
        self.block3 = SpikeBlock(192, 384, tau)

        self.refine = nn.Sequential(
            nn.Conv2d(
                384, 384,
                kernel_size=3,
                padding=1,
                groups=384,
                bias=False
            ),
            nn.BatchNorm2d(384),

            nn.Conv2d(
                384, 256,
                kernel_size=1,
                bias=False
            ),
            nn.BatchNorm2d(256),

            neuron.LIFNode(
                tau=tau,
                v_threshold=1.0,
                surrogate_function=surrogate.ATan(alpha=2.0),
                step_mode='s'
            )
        )

        self.gap = nn.AdaptiveAvgPool2d(1)

    def forward(self, x):
        x = self.stem(x)
        x = self.block1(x)
        x = self.block2(x)
        x = self.block3(x)
        x = self.refine(x)
        x = self.gap(x).flatten(1)

        return x


# 5. Dual-Branch NeurALLNet
class DualBranchNeurALLNet(nn.Module):
    def __init__(self, num_classes=11, time_steps=4):
        super().__init__()

        self.T = time_steps

        # Clinical branch
        self.clinical_branch = NeurALLFeatureExtractor()

        # Dermoscopic branch
        self.dermoscopic_branch = NeurALLFeatureExtractor()

        # Feature fusion + classifier
        self.classifier = nn.Sequential(
            nn.Linear(512, 128, bias=False),
            nn.BatchNorm1d(128),

            neuron.LIFNode(
                tau=2.0,
                v_threshold=1.0,
                surrogate_function=surrogate.ATan(alpha=2.0),
                step_mode='s'
            ),

            nn.Dropout(0.3),

            nn.Linear(128, num_classes, bias=False)
        )

    def forward(self, clinical, dermoscopic):

        output_sum = None

        for t in range(self.T):

            # Reset spiking states at each timestep
            functional.reset_net(self)

            clinical_features = self.clinical_branch(clinical)

            dermoscopic_features = self.dermoscopic_branch(
                dermoscopic
            )

            # Concatenate both feature vectors
            fused_features = torch.cat(
                (clinical_features, dermoscopic_features),
                dim=1
            )

            output = self.classifier(fused_features)

            if output_sum is None:
                output_sum = output
            else:
                output_sum = output_sum + output

        return output_sum / self.T


print("All architecture classes defined successfully.")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = DualBranchNeurALLNet(
    num_classes=11,
    time_steps=4
).to(device)

total_params = sum(
    p.numel() for p in model.parameters()
)

trainable_params = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print("Device:", device)
print("Model:", model.__class__.__name__)
print("Time steps:", model.T)
print("Total parameters:", f"{total_params:,}")
print("Trainable parameters:", f"{trainable_params:,}")

print("\nGPU allocated:",
      round(torch.cuda.memory_allocated() / 1024**3, 2), "GB")
```

```python
import copy
import torch
import os
import time
import numpy as np

# Training configuration
EPOCHS = 50
PATIENCE = 10
BEST_VAL_ACC = 0.0
best_epoch = 0
epochs_without_improvement = 0

CHECKPOINT_DIR = "/kaggle/working/checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

BEST_MODEL_PATH = os.path.join(
    CHECKPOINT_DIR, "best_neurallnet_milk10k.pth"
)

history = {
    "train_loss": [],
    "train_acc": [],
    "val_loss": [],
    "val_acc": []
}

for epoch in range(EPOCHS):

    start_time = time.time()

    # ---------------- TRAINING ----------------
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for clinical, dermoscopic, labels in train_loader:

        clinical = clinical.to(device)
        dermoscopic = dermoscopic.to(device)
        labels = labels.to(device)

        optimizer.zero_grad(set_to_none=True)

        outputs = model(clinical, dermoscopic)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item() * labels.size(0)

        predictions = outputs.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    train_loss = running_loss / total
    train_acc = 100 * correct / total

    # ---------------- VALIDATION ----------------
    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for clinical, dermoscopic, labels in val_loader:

            clinical = clinical.to(device)
            dermoscopic = dermoscopic.to(device)
            labels = labels.to(device)

            outputs = model(clinical, dermoscopic)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * labels.size(0)

            predictions = outputs.argmax(dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    val_loss = val_running_loss / val_total
    val_acc = 100 * val_correct / val_total

    # Learning rate update
    scheduler.step()

    # Store history
    history["train_loss"].append(train_loss)
    history["train_acc"].append(train_acc)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)

    # ---------------- CHECKPOINT ----------------
    if val_acc > BEST_VAL_ACC:

        BEST_VAL_ACC = val_acc
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        torch.save({
            "epoch": epoch + 1,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_acc": BEST_VAL_ACC,
            "history": history,
        }, BEST_MODEL_PATH)

        status = "Best model saved"

    else:
        epochs_without_improvement += 1
        status = f"No improvement: {epochs_without_improvement}/{PATIENCE}"

    elapsed = time.time() - start_time

    print(f"\nEpoch [{epoch+1}/{EPOCHS}]")
    print(f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
    print(f"Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
    print(f"Learning Rate: {optimizer.param_groups[0]['lr']:.6f}")
    print(f"Time: {elapsed:.1f}s | {status}")

    # Early stopping
    if epochs_without_improvement >= PATIENCE:
        print("\nEarly stopping triggered.")
        break

print("\n========== TRAINING COMPLETE ==========")
print("Best validation accuracy:", round(BEST_VAL_ACC, 2), "%")
print("Best epoch:", best_epoch)
print("Checkpoint:", BEST_MODEL_PATH)
```

```python
from sklearn.metrics import classification_report, accuracy_score
import torch

# Load best checkpoint
checkpoint = torch.load(
    BEST_MODEL_PATH,
    map_location=device,
    weights_only=True
)

model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

y_true = []
y_pred = []

with torch.no_grad():
    for clinical, dermoscopic, labels in test_loader:

        clinical = clinical.to(device)
        dermoscopic = dermoscopic.to(device)

        outputs = model(clinical, dermoscopic)
        predictions = outputs.argmax(dim=1)

        y_true.extend(labels.numpy())
        y_pred.extend(predictions.cpu().numpy())

# Final metrics
test_accuracy = accuracy_score(y_true, y_pred)

print("Best checkpoint epoch:", checkpoint["epoch"])
print("Best validation accuracy:", checkpoint["best_val_acc"])
print("\nTest Accuracy:", round(test_accuracy * 100, 2), "%")

print("\nClassification Report:\n")

print(classification_report(
    y_true,
    y_pred,
    labels=list(range(len(classes))),
    target_names=classes.tolist(),
    zero_division=0
))
```
